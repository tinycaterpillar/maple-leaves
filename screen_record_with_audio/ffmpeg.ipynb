```python
from subprocess import call
import argparse
import time
```

### 녹화 영역 지정

```python
import pyautogui
import winsound as sd

def beepsound():
    fr = 2000    # range : 37 ~ 32767
    du = 200     # 1000 ms ==1second
    sd.Beep(fr, du) # winsound.Beep(frequency, duration)

time.sleep(2)
x1, y1 = pyautogui.position()
beepsound()
time.sleep(2)
x2, y2 = pyautogui.position()
beepsound()

print(f'args.offset_x = {x1}')
print(f'args.offset_y = {y1}')
print(f'args.width = {x2-x1}')
print(f'args.height = {y2-y1}')
```

```python
parser = argparse.ArgumentParser()
# parser.add_argument('-n', type=int)
args = parser.parse_args("")

# record(ffmpeg)
args.duration = "00:00:43"
args.file_name = "bellman-ford"

args.offset_x = 653
args.offset_y = 171
args.width = 358
args.height = 462
args.show_mouse = False
args.show_region = True
```

### record screen without audio

```python
time.sleep(3)
cmd = f"ffmpeg -y -f gdigrab -t {args.duration} -framerate 30 -offset_x {args.offset_x} -offset_y {args.offset_y} -s {args.width}x{args.height} -show_region {int(args.show_region)} -draw_mouse {int(args.show_mouse)} -i desktop {args.file_name}.mp4"
print(cmd)
call(cmd)
```

### record screen with audio

```python
time.sleep(3)
cmd = f'ffmpeg -y -f dshow -t {args.duration} -i audio="virtual-audio-capturer" -rtbufsize 100M -f gdigrab -t {args.duration} -framerate 30 -offset_x {args.offset_x} -offset_y {args.offset_y} -s {args.width}x{args.height} -show_region {int(args.show_region)} -probesize 10M -draw_mouse {int(args.show_mouse)} -i desktop -c:v libx264 -r 30 -preset veryfast -tune zerolatency -crf 25 -pix_fmt yuv420p {args.file_name}.mp4' 
print(cmd)
call(cmd)
```
